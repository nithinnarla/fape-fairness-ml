# FAPE, Group-Size Check for COMPAS and Folktables

Neither intervention script sets a minimum group size when it computes demographic parity difference or equalized odds difference, and both test splits hold race groups with a handful of records: 7 Asian defendants and 1 Native American defendant in COMPAS, 5 Alaska Native and 25 American Indian and Alaska Native respondents in Folktables. The script rebuilds each domain's intervention setup and prints, for every model, each group's selection rate before and after the demographic parity constraint, then both metrics measured over all groups and over groups with at least 30 test records. No reported result changes; the output shows which groups set the values.

Code: `src/group_size_check.py`. Section 6.2 of the paper and Table 6 of `docs/cross_domain_results_table.md` use this output.
