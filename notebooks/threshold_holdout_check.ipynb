# FAPE, Threshold-Fitting Check for Random Forest

Every intervention script gives ThresholdOptimizer the training split, so the group thresholds are chosen on the records the model was fit to. The script runs that design next to a held-out one, where the model trains on 75% of the training split and the thresholds are chosen on the other 25%, for Student (Math) and MEPS, the two evaluations where random forest worsens under a constraint and small groups do not explain it. For each model it prints the training accuracy, then DPD and EOD before and after each constraint under both designs, all scored on the same test split.

Code: `src/threshold_holdout_check.py`. Section 6.4 of the paper and Table 7 of `docs/cross_domain_results_table.md` use this output.
